**#DATA TYPES**

* Python int: can represent 10**308 exactly.

* Python float: can represent values up to about 1.797 × 10^308.

```python
# Integer
print(4)
# 1 * 10 ^308 == 1e308
print(1e308)
print(type(4))
print(type(1e308))

#4 → int (integer)

#1e308 → float (floating-point number)

#1e308 means 1 × 10³⁰⁸
```

* 4 → int (integer)

* 1e308 → float (floating-point number)

* 1e308 means 1 × 10³⁰⁸
* In Python, ^ does not mean exponentiation. It means bitwise XOR.

   For powers, Python uses **:

* print(2**3)   # 8
* print(2^3)    # 1







```python
# Integer
print(4)
# 1 * 10 ^309 == 1e309
print(1e309)
```

**Why inf?**

* Python's float is usually a 64-bit IEEE 754 floating-point number, whose maximum finite value is approximately: 1.7976931348623157 × 10^308

**Therefore:**

* 1e308   # works → 1e+308
* 1e309   # too large → inf

**So the important distinction is:**

| Expression | Type | Result |
|---|---|---|
| `1e308` | `float` | `1e+308` |
| `1e309` | `float` | `inf` |
| `10**308` | `int` | exact integer |
| `10**309` | `int` | exact integer |

1e309 doesn't mean Python integers stop at 308. It means Python's float cannot represent a finite number that large.

```python
#x = 10**309
#x = 10**409
x = 10**4300

print(x)
print(type(x))
```

* Yes — this is a very useful Python behavior to understand. Your calculation succeeds, but print(x) fails.

* **What happened?**
* x = 10**4300

* print(x)        # ❌ ValueError
* print(type(x))  # would be int

* 10**4300 creates an integer containing 4301 decimal digits:

* 10^4300 = 1 followed by 4300 zeros

* Python can create and store this huge int, but Python 3.11+ has a security limit on converting very large integers to decimal strings.
The default limit is 4300 digits.

```python
# x = 10**4300 #it is produce 4301 digits so it gave the error
x = 10**4299

print(type(x))
print(len(str(x)))
```

**#Variables**

* A variable is a name used to store a value.

* Variables are created using the = sign.

* Python does not require declaring the data type of a variable.

**Dynamic Typing**
* Dynamic typing means the type of a variable is determined at runtime.

* You don't need to declare the data type explicitly.

* A variable can hold different types of values during the program.

* Python is dynamically typed.

```python
x = 10
print(x)
name = "Hello"
print(name)
```

**Static Typing**
* Static typing means the variable's type is determined before/runtime checking and generally cannot change without an explicit conversion or incompatible assignment is rejected.

* The data type is usually declared explicitly or inferred and then checked.

* Examples of statically typed languages: C, C++, Java, C#.

```python
#int x = 10;
#x = "Hello";  // Error
```

## Key Difference

| Dynamic Typing | Static Typing |
|---|---|
| Type checked at runtime | Type checked before execution/compile time |
| No need to declare type explicitly | Type is declared or inferred and checked |
| Python, JavaScript | C, C++, Java |
| More flexible | More strict |


## Dynamic Binding vs Static Binding

| Dynamic Binding | Static Binding |
|---|---|
| Binding is done at runtime | Binding is done at compile time |
| Also called late binding | Also called early binding |
| Method to execute is decided during program execution | Method to execute is decided before execution |
| More flexible | Less flexible |
| Commonly associated with method overriding | Commonly associated with method overloading |
| Example: Python supports dynamic method dispatch | Example: Java supports static binding for overloaded methods |


```python
#Dynamic Binding
x = "Kuvar"
print(x)
x = 7
print(x)
```
